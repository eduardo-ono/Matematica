__EsPCEx-2022-q11__. &nbsp; Sejam $\lambda$ um parâmetro real e $\varepsilon$ o sistema linear abaixo, com incógnitas $a$, $b$ e $c$.

$$
\varepsilon = \left\{
  \begin{matrix}
  \ \lambda a & - & b & + & c & = & 3 \\
  \ a & + & \lambda b & + & c & = & \lambda \\
  \ a & + & b & + & c & = & 1
  \end{matrix}
\right.
$$

&emsp; É correto afirmar que

<sample>a)</sample>&nbsp; $\varepsilon$ será possível e determinado se $\,\lambda = 0\,$ ou $\,\lambda = 1\,$.<br>
<sample>b)</sample>&nbsp; $\varepsilon$ será possível e indeterminado se, e somente se, $\,\lambda = 0\,$ ou $\,\lambda = 1\,$.<br>
<sample>c)</sample>&nbsp; $\varepsilon$ será impossível se $\,\lambda = -1\,$ ou $\,\lambda = 0\,$. &check;<br>
<sample>d)</sample>&nbsp; $\varepsilon$ será possível e indeterminado se $\,\lambda = -1\,$ ou $\,\lambda = 0\,$.<br>
<sample>e)</sample>&nbsp; $\varepsilon$ será impossível se, e somente se, $\,\lambda = 0\,$ ou $\,\lambda = 1\,$.<br>

__exr-2020-11-03__. &nbsp; Verifique que as matrizes a seguir são invertíveis e determine suas respectivas inversas.

$$
A =
\begin{pmatrix}
1 & 5 \\
3 & 2
\end{pmatrix}
\quad \text{e} \quad
B = 
\begin{pmatrix}
1 & 2 & 1\\
3 & 2 & 4 \\
1 & -1 & 2
\end{pmatrix}
$$

&emsp; Resp.:

$$
A^{-1} =
\begin{pmatrix}
\cfrac{-2}{13} & \cfrac{5}{13} \\
\cfrac{3}{13} & \cfrac{-1}{13}
\end{pmatrix}
\quad \text{e} \quad
B^{-1} =
\begin{pmatrix}
-8 & 5 & -6 \\
2 & -1 & 1 \\
5 & -3 & 4
\end{pmatrix}
$$
