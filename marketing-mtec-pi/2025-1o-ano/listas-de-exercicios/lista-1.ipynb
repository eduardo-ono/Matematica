&emsp; Nos exercícios 1 e 2, considere o conjunto $A$ descrito da seguinte forma:

$$
A = \set{x \in \mathbb{N} \mid x \text{ é um algarismo (dígito) do seu RM.}}
$$

&emsp; Por exemplo, caso o seu RM seja 25123, então $A = \set{2, 5, 1, 3}$.

__EXERCÍCIO 1__.

&emsp; Sejam dados os conjuntos $\ U = \set{1, 3, 5, 7, 9}\ $ e $\ W = \set{0, 2, 4, 6, 8}\ $. Determine:

$$
\begin{align*}
i) \quad & A \qquad\qquad\qquad \\
ii) \quad & A \cup U \\
iii) \quad & A \cap U \\
iv) \quad & A \cup W \\
v) \quad & A \cap W \\
vi) \quad & U \cap W \\
vii) \quad & A - U \\
viii) \quad & U - A \\
ix) \quad & W - A \\
x) \quad & \complement_W^A
\end{align*}
$$


__EXERCÍCIO 2__.

&emsp; Dado um conjunto $B$, chama-se __conjunto das partes__ de $B$ – notação $\wp(B)$ – aquele que é formado por todos os subconjuntos de $B$, ou seja,

$$
\wp(B) = \{X \mid X \subset B\}.
$$

&emsp; Lembrando que o conjunto vazio é subconjunto de qualquer conjunto, determine:

i) &nbsp; $\,\wp(A)\,$

ii) &nbsp; $n(\wp(A))$

&emsp; Obs.: $n(\wp(A))$ é a cardinalidade de $\wp(A)$.

&emsp; Nos exercícios 3 e 4, considere o número natural $N$ como sendo a soma de todos os algarismos de seu RM. Por exemplo, se seu RM for igual a 25123, então $N = 2 + 5 + 1 + 2 + 3 = 13$.

__EXERCÍCIO 3__.

&emsp; Uma pesquisa sobre preferências de idiamas foi realizada com 1000 alunos de uma escola. Nessa pesquisa, 600 disseram gostar de português, 400 de inglês e 300 de espanhol. Além disso, 150 alunos disseram gostar de português e inglês, 100 de português e espanhol e 50 de inglês e espanhol. Por fim, $N$ alunos disseram gostar dos três idiomas.

&emsp; Usando um diagrama de Venn, determine:

i) Quantos alunos gostam somente de português?

ii) Quantos alunos gostam somente de inglês?

iii) Quantos alunos gostam somente de espanhol?

iv) Quantos alunos gostam de exatamente um dos idiomas?

v) Quantos alunos gostam de exatamente dois dos idiomas?

vi) Quantos alunos gostam de pelo menos um dos idiomas?

vii). Quantos alunos preferem nenhum dos idiomas?

__EXERCÍCIO 4__.
