__EsPCEx-2016-q20__. &nbsp; Corta-se de uma circunferência de raio 4 cm, um setor circular de ângulo $\cfrac{\pi}{2}$ rad (ver desenho ilustrativo), onde o ponto $C$ é centro da circunferência. Um cone circular reto é construído a partir desse setor circular ao se juntar os rais $\,CA\,$ e $\,CB$. O volume desse cone, em cm<sup>3</sup>, é igual a

<sample>C)</sample>&nbsp; $\cfrac{\sqrt{15}}{3}\,\pi$ &check;<br><br>
