&emsp; Mostre que os zeros da função $\,f(x) = ax^2 + bx + c\,$, com $\,a \neq 0\,$, são dados por:

$$
x = \cfrac{-b \pm \sqrt{b^2 - 4ac}}{2a}\,\cdot
$$

<hr>

&emsp; Mostre que as coordenadas do vértice de uma parábola são dadas por:

$$
x_V = \cfrac{-b}{2a} \quad \text{e} \quad y_V = \cfrac{-\Delta}{4a}\,,
$$

onde $\,\Delta = b^2 - 4ac\,$.

<hr>

&emsp; Mostre que a função quadrática $\,f(x) = ax^2 + bx + c\,$, com $\,a \neq 0\,$, pode ser fatorada na forma $\,f(x) = a(x - x_1)(x - x_2)$.

<hr>

&emsp; Construa o gráfico cartesiano das seguintes funções definidas em $\mathbb{R}$:

a) $y = x^2 - 2x - 3$<br><br>
b) $y = 4x^2 - 10x + 4$<br><br>
c) $y = -x^2 + \cfrac{1}{2}\,x + \cfrac{1}{2}$<br><br>
d) $y = -3x^2 + 6x - 3$<br><br>
e) $y = x^2 - 3x + \cfrac{9}{4}$<br>

<hr>

&emsp; A função $\,f: \mathbb{R} \longrightarrow \mathbb{R}$ tem como gráfico uma parábola e satisfaz $\,f(x + 1) - f(x) = 6x - 2\,$, para todo número real $x$. Determine o valor de $x$ tal que $f(x)$ seja mínimo.

<hr>

&emsp; Uma empresa fabrica um único produto na qual o lucro $L(x)$, em milhares de reais, é descrito pela função $\,L(x) = -2x^2 + 16x - 20\,$, onde $x$ representa o preço unitário do produto. Qual deve ser o preço do produto para que a empresa obtenha o maior lucro possível?

<hr>

&emsp; A função quadrática $\,f(x) = 3x^2 - 6x + k\,$ não tem raízes reais. Qual o menor valor inteiro para $k$?

<hr>

&emsp; Determine a função $g(x)$ cujo gráfico é o simétrico do gráfico da função $\,f(x) = 2x - x^2\,$ em relação à reta $\,y = 3\,$. Esboce o gráfico.

<hr>

&emsp; As duas raízes da equação $\,x^2 - 63x + k = 0\,$ na incógnita $x$ são números inteiros e primos. Determine o total de valores distintos que $k$ pode assumir.

<hr>

&emsp; Uma agência de turismo, que promove excursões em um parque florestal, atrai 40 clientes quando cobra $\text{R\$ } 50,00$ por um passeio de 4 horas. Dados estatísticos mostram que, para cada redução de $\text{R\$ } 5,00$ nesse preço, há um aumento de 10 no número de pessoas que participam de um passeio. Com base nessas informações, estime o preço por pessoa para que essa empresa obtenha a receita máxima com um desses passeios.

<hr>

&emsp; Pedro escreveu o número 10 como soma de duas parcelas inteiras positivas e cujo produto é o maior possível. Determine o valor desse produto.

<hr>

&emsp; A função quadrática $\,f(x) = 16x - x^2\,$ está definida no domínio dado pelo intervalo $[0, 7]$. Determine a imagem máxima de $f$.

<hr>

&emsp; Uma partícula tem sua trajetória retilínea definida pela função que relaciona a distância $S$, em metros, da partícula a um ponto fixo e o tempo $t$, em segundos, dada por $\,S(t) = 45 + 40t - 5t^2\,$. Determine quantos metros foram percorridos entre 3 segundos e 6 segundos a partir do instante inicial zero.

<hr>

&emsp; Em um terreno plano horizontal, está fixado um mastro vertical com 13,5 metros de altura. Do topo do mastro, é lançado um projétil, descrevendo uma trajetória de modo que sua altura, em relação ao terreno, é uma função quadrática de sua distância à reta que contém o mastro. O projétil alcança a altura de 16 metros, quando essa distância é
de 3 metros, e atinge o solo, quando a distância é de 27 metros. Determine, em metros, a altura máxima alcançada pelo projétil.

<hr>

&emsp; O transporte aéreo de pessoas entre duas cidades $A$ e $B$ é feito por uma única companhia em um único voo diário. O avião utilizado tem 180 lugares, e o preço da passagem $p$ relaciona-se com o número $x$ de passageiros por dia pela relação $\,p = 300 - 0,75x\,$. Determine a receita máxima possível por viagem.

<hr>

&emsp; Um supermercado vende 400 pacotes de 5 kg de uma determinada marca de açúcar por semana. O preço de cada pacote é $\text{R\$ } 6{,}00$, e o lucro do supermercado, em cada pacote vendido, é de $\text{R\$ } 2{,}00$. Se for dado um desconto de $x$ reais no preço do pacote de açúcar, o lucro por pacote terá uma redução de $x$ reais, mas, em compensação, o supermercado aumentará sua venda em 400x pacotes por semana. Nestas condições, calcule:

a) O lucro desse supermercado em uma semana, caso o desconto dado seja de $\text{R\$ } 1{,}00$.<br>
b) O preço do pacote de açúcar para que o lucro do supermercado seja máximo, no período considerado.<br>

<hr>

&emsp; Joãozinho, ao falecer, deixou uma herança de $\text{R\$ } 200\,000{,}00$ para ser distribuída, de maneira equitativa, entre os seus $x$ filhos. No entanto, três desses filhos renunciaram às suas respectivas partes nessa herança, fazendo com que os demais $x - 3$ filhos, além do que receberiam normalmente, tivessem um adicional de $\text{R\$ } 15\,000{,}00$ em suas respectivas partes dessa herança. Determine o número $x$ de filhos de Joãozinho.

<hr>

&emsp; Um restaurante a quilo vende 100 kg de comida por dia, a $\text{R\$ } 15{,}00$ o quilograma. Uma pesquisa de opinião revelou que, a cada real de aumento no preço do quilo, o restaurante deixa de vender o equivalente a 5 kg de comida. Responda às perguntas abaixo, supondo corretas as informações da pesquisa e definindo a receita do restaurante como o valor total pago pelos clientes.

a) Em que caso a receita do restaurante será maior: se o preço subir para $\text{R\$ } 18{,}00$/kg ou para $\text{R\$ } 20{,}00$/kg?<br>
b) Formule matematicamente a função $f(x)$, que fornece a receita do restaurante como função da quantia $x$, em reais, a ser acrescida ao valor atualmente cobrado pelo quilo da refeição.<br>
c) Qual deve ser o preço do quilo da comida para que o restaurante tenha a maior receita possível?<br>

<hr>

&emsp; Um grupo de $x$ amigos decidiram fazer uma festa cujo custo é de $\text{R\$ } 3\,250{,}00$. Alguns dias depois, mais três pessoas se juntaram ao grupo, formando um novo grupo com $x + 3$ pessoas. Ao fazer a divisão do valor da festa pelo número de pessoas que estão compondo o novo grupo, verificou-se que cada pessoa pagaria $\text{R\$ } 75{,}00$ a menos do que o inicialmente programado para cada um no primeiro grupo. Determine o número de pessoas que formavam o primeiro grupo.

<hr>

&emsp; Durante o mês de janeiro, uma papelaria obteve um total de $\text{R\$ } 900,00$ pelas vendas de um compasso escolar. Com a chegada do mês de fevereiro, a papelaria decidiu dar um desconto para estimular as vendas, baixando o preço desse compasso em $\text{R\$ } 10,00$. Com isso, vendeu em fevereiro 5 compassos a mais do que em janeiro, obtendo um total de $\text{R\$ } 1\,000,00$ pelas vendas de fevereiro. Determine o preço pelo qual esse compasso foi vendido em janeiro.

<hr>

&emsp; Considere a equação:

$$
\sqrt{x^2 - p} + 2 \sqrt{x^2 -1} = x
$$

a) Para que valores do parâmetro real $p$ a equação admite raízes reais?<br>
b) Determine todas essas raízes reais.<br>

<hr>

&emsp; Uma empresa de tecnologia lançou um novo modelo de drone e deseja calcular a altura máxima, em metros, que ele pode atingir, que é modelada pela função do segundo grau $\,h(t) = - 5t² + 20t\,$, onde $t$ é o tempo em segundos após o drone começar a subir. Qual é a altura máxima que o drone atinge?

<hr>

&emsp; Uma função quadrática $f(x)$ é tal que $\,f(0) = 4\,$, $\,f(1) = -3\,$ e $\,f(-1) = 15\,$. Determine os zeros dessa função.
