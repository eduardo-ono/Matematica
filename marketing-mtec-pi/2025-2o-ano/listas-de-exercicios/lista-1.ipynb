&emsp; Nos exercícios 1 a 3, considere o número $N$ como sendo a soma de todos os algarismos (dígitos) do seu RM (Registro de Matrícula). Por exemplo, se o seu RM for 025123, então $N = 2 + 5 + 1 + 2 + 3 = 13$.

__EXERCÍCIO 1__.

&emsp; Determine o valor de $x$ na equação a seguir:

$$
\log^2 x + N \cdot \log x^2 + N = 0
$$

&emsp; Obs.: $\,\log^2 x = (\log x)^2\,$.

__EXERCÍCIO 2__.

&emsp; Considere que a taxa da inflação de um país se manteve constante durante os últimos 12 meses. Sabendo-se que, nesse mesmo país, a taxa da inflação anual foi de $N \%$ a.a. (ao ano), determine a taxa da inflação mensal nesse período.
