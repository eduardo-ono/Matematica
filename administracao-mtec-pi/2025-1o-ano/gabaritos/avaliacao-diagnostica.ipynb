<h2>Questão 4.</h2>

__Método de resolução 1:__

$$
\begin{align*}
(a + b + c)^2 &= [(a + b) + c]^2 \\[8pt]
&= (a + b)^2 + 2 \, (a + b) \, c + c^2 \\[8pt]
&= (a + b)^2 + 2 c \, (a + b) + c^2 \\[8pt]
&= (a^2 + 2ab + b^2) + 2ac + 2bc + c^2 \\[8pt]
&= a^2 + b^2 + c^2 + 2ab + 2ac + 2bc \\
\end{align*}
$$

__Método de resolução 2:__

$$
\begin{align*}
\text{Fazendo } m = a + b, \text{ temos:} \\
(a + b + c)^2 &= (m + c)^2 \\[8pt]
&= m^2 + 2 m c + c^2 \\[8pt]
&= (a + b)^2 + 2 c \, (a + b) + c^2 \\[8pt]
&= (a^2 + 2ab + b^2) + 2ac + 2bc + c^2 \\[8pt]
&= a^2 + b^2 + c^2 + 2ab + 2ac + 2bc \\
\end{align*}
$$

<h2>Questão 5.</h2>


1º Passo: Classificar o triângulo quanto aos lados (equilátero, isóceles, escaleno) e quanto aos ângulos (retângulo, acutângulo, obtusângulo)

$$
\begin{align*}
15^2 &= 9^2 + 12^2 \\
225 &= 81 + 144 \text{ (V)}
\end{align*}
$$

Portanto, pela recíproca do Teorema de Pitágoras, o triângulo é retângulo.

Resolução 1:

&emsp; Se o triângulo é retâgulo, a área do triângulo é dada por $A = \cfrac{9 \cdot 12}{2}$.

&emsp; A mesma área pode também ser obtida por $A = \cfrac{15 \cdot h}{2}$.

&emsp; Igualando as duas áreas, temos:

$$
\begin{align*}
\cfrac{15 \cdot h}{2} &= \cfrac{9 \cdot 12}{2} \\[10pt]
h &= \cfrac{9 \cdot 12}{15} \\[10pt]
h &= \cfrac{9 \cdot 4}{5} \\[10pt]
h &= 7{,}2
\end{align*}
$$
