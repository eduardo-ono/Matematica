__ITA-2023-2ª-fase-q02__. &nbsp; Considere as seguintes matrizes:

$$
A = 
\begin{bmatrix}
1 & -2 \\
-2 & 1
\end{bmatrix}
, \quad
B =
\begin{bmatrix}
0 & 6 \\
6 & 0
\end{bmatrix}
\quad \text{e} \quad
C =
\begin{bmatrix}
3 & 3 \\
3 & 3
\end{bmatrix}
.
$$

&emsp; Determine os números $\,\alpha \in \mathbb{R}\,$ tais que a matriz $\,M = \alpha^2 A + \alpha B + C\,$ é invertível.

__ITA-2015-q28__. &nbsp; Sejam $\alpha$ e $\beta$ números reais não nulos. Determine os valores de $b$, $c$, $d$, bem como a relação entre $\alpha$ e $\beta$ para que ambos os sistemas lineares $S$ e $T$ a seguir sejam compatíveis indeterminados.

$$
S
\left\{
\begin{matrix}
\ 2x & + & by & = & \alpha \\
\ cx & + & y & = &\beta
\end{matrix}
\right.
\qquad\qquad
T
\left\{
\begin{matrix}
\ cx & + & 3y & = & \alpha \\
\ 4x & + & dy & = &\beta
\end{matrix}
\right.
$$
